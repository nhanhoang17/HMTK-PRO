```python

le_user = LabelEncoder()
df_new['user_id_enc'] = le_user.fit_transform(df_new['user_id'])

unique_users = sorted(df_new['user_id'].unique())
unique_tracks = sorted(df_new['track_id'].unique())

user2idx = {u: i for i, u in enumerate(unique_users)}
track2idx = {t: i for i, t in enumerate(unique_tracks)}


idx2user = {i: u for u, i in user2idx.items()}
idx2track = {i: t for t, i in track2idx.items()}

df_new['user_id_enc'] = df_new['user_id'].map(user2idx)
df_new['track_id_enc'] = df_new['track_id'].map(track2idx)
```

```python
from scipy.sparse import csr_matrix
import numpy as np


n_users = df_new['user_id_enc'].max() + 1    # 66299
n_songs = df_new['track_id_enc'].max() + 1   # 149993


sparse_matrix = csr_matrix(
    (df_new['Play count'].values,
     (df_new['user_id_enc'].values, df_new['track_id_enc'].values)),
    shape=(n_users, n_songs)
)

print(f"Shape       : {sparse_matrix.shape}")
print(f"Non-zeros   : {sparse_matrix.nnz:,}")
print(f"Sparsity    : {1 - sparse_matrix.nnz / (n_users * n_songs):.6f}")
```

```python
#Kiểm tra độ sparse 
nnz = sparse_matrix.count_nonzero()

rows, cols = sparse_matrix.shape

sparsity = 1 - (nnz / (rows * cols))

print(f"Sparsity: {sparsity:.4f}")
```

```python
#Bước 8 
from scipy.sparse.linalg import svds
from scipy.sparse import csr_matrix
import numpy as np
```

```python
# ── 8.1  Chuẩn hoá Play count → log scale
# Play count có thể rất lệch (vài user nghe 1 bài hàng trăm lần)
# Log giúp giảm ảnh hưởng của outliers, tương tự như implicit feedback
log_matrix = sparse_matrix.copy().astype(np.float32)
log_matrix.data = np.log1p(log_matrix.data)   # log(1 + play_count)
```

```python
# ── 8.2  Truncated SVD ─
# R ≈ U · Σ · Vt
# - U   : (n_users, K)  — user latent factors
# - Σ   : (K,)          — singular values (độ "quan trọng" mỗi chiều)
# - Vt  : (K, n_songs)  — song latent factors (transpose)
#
# k=64 → mỗi user và song được biểu diễn bằng 64 chiều ẩn
 
K = 64
 
print(f"Đang chạy Truncated SVD với K={K} chiều ẩn...")
print(f"Ma trận đầu vào: {log_matrix.shape[0]:,} users × {log_matrix.shape[1]:,} songs")
 
U, sigma, Vt = svds(log_matrix, k=K)
 
# svds trả về theo thứ tự tăng dần → đảo ngược để chiều quan trọng nhất ở đầu
U      = U[:, ::-1]          # (66299, 64)
sigma  = sigma[::-1]         # (64,)
Vt     = Vt[::-1, :]         # (64, 173980)
```

```python
# ── 8.3  Tạo embedding vectors
# Nhân với căn sigma để scale đúng mức — cách chuẩn trong SVD-based MF
# user_embeddings[u]  = vector 64 chiều của user u
# song_embeddings[i]  = vector 64 chiều của song i
 
sqrt_sigma = np.sqrt(sigma)                  # (64,)
 
user_embeddings = U * sqrt_sigma             # (66299, 64)
song_embeddings = (Vt.T) * sqrt_sigma        # (173980, 64)
```

```python
# ── 8.4  Kiểm tra kết quả 
print("\n SVD hoàn tất!")
print(f"   user_embeddings : shape = {user_embeddings.shape}  | dtype = {user_embeddings.dtype}")
print(f"   song_embeddings : shape = {song_embeddings.shape}  | dtype = {song_embeddings.dtype}")
print(f"   Sigma (top 5)   : {sigma[:5].round(2)}")
 
# Phần trăm variance được giải thích bởi K chiều
explained = (sigma**2).sum() / (log_matrix.data**2).sum() * 100
print(f"   Variance giải thích được: {explained:.2f}%")
```

```python
# ── 8.5  Ví dụ: lấy embedding của user/song bất kỳ 
sample_user_id  = df_new['user_id'].iloc[0]       # user đầu tiên trong df
sample_track_id = df_new['track_id'].iloc[0]      # track đầu tiên trong df
 
u_idx = user2idx[sample_user_id]
t_idx = track2idx[sample_track_id]
 
print(f"\n Ví dụ embedding:")
print(f"   User  {sample_user_id}  → index {u_idx} → vector (5 chiều đầu): {user_embeddings[u_idx, :5].round(4)}")
print(f"   Track {sample_track_id} → index {t_idx} → vector (5 chiều đầu): {song_embeddings[t_idx, :5].round(4)}")
```

```python
# ── 8.6  Kiểm tra dự đoán: score user-song 
# Dự đoán "độ phù hợp" = tích vô hướng user vector · song vector
pred_score = np.dot(user_embeddings[u_idx], song_embeddings[t_idx])
print(f"\n   Predicted score (user × song dot product): {pred_score:.4f}")
```

```python
# ── 8.7  Recommend top-N songs cho 1 user 
def recommend_songs(user_id, top_n=10):
    """
    Trả về top_n song được dự đoán là phù hợp nhất với user_id.
    Loại bỏ những bài user đã nghe.
    """
    u_idx = user2idx[user_id]
 
    # Tính score với tất cả songs: (64,) · (173980, 64).T → (173980,)
    scores = song_embeddings @ user_embeddings[u_idx]
 
    # Lấy index các bài user ĐÃ nghe để loại ra
    listened = set(sparse_matrix[u_idx].indices)
 
    # Sắp xếp giảm dần, bỏ những bài đã nghe
    ranked = np.argsort(-scores)
    rec_indices = [i for i in ranked if i not in listened][:top_n]
 
    # Map về track_id gốc
    return [idx2track[i] for i in rec_indices]
 
```

```python
# Test thử
sample_recs = recommend_songs(sample_user_id, top_n=5)
print(f"\n Top-5 gợi ý cho user {sample_user_id}:")
for rank, track_id in enumerate(sample_recs, 1):
    print(f"   {rank}. {track_id}")
```
