```python
import kagglehub
import shutil
from pathlib import Path

# Download dataset từ Kaggle
path = kagglehub.dataset_download("chelseapower/nowplayingrs")

print("Dataset downloaded to:", path)

# Thư mục muốn lưu ở ổ D
destination = Path(r"D:\KaggleDatasets\nowplayingrs")


if destination.exists():
    shutil.rmtree(destination)

# Copy dataset sang ổ D
shutil.copytree(path, destination)

print("Dataset copied to:", destination)
```

```python
import os
import pandas as pd

path = r"D:\KaggleDatasets\nowplayingrs"

csv_files = []

for root, dirs, files in os.walk(path):
    for file in files:
        if file.lower().endswith(".csv"):
            csv_files.append(os.path.join(root, file))

print("CSV files found:")
for i, file in enumerate(csv_files):
    print(i, file)
```
